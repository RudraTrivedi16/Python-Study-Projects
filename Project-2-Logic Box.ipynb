```python
print("""Welcome to A fun program Where you can 
Generate Patterns 
       OR
Analyze Numbers""")
while True:
    print("\n")
    print("""Select one option from below:
    1. Generate a Pattern.
    2. Analyze a Range of numbers.
    3. Exit.""")
    choice=int(input("Enter your choice: "))
    print("\n")
    if choice==1:
        rows=int(input("Enter the number of rows you want to print: "))
        pattern=input("Enter the pattern you want to print: ")
        print("\n")
        for i in range(1,rows+1):
            print(pattern*i)
            i+=1
    elif choice==2:
        first=int(input("Enter the start of the range: "))
        second=int(input("Enter the end of the range: "))
        print("\n")
        for j in range(first,second+1):
            if j%2==0:
                print(f"{j} is Even number.")
            elif j%2!=0:
                print(f"{j} is Odd number.")
    elif choice==3:
        break
    else:
        print("Wrong Input Detected. Try again.")
        continue
```
