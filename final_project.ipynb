```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

class RetailAnalyzer:
    def __init__(self):
        self.df = None

    def load_data(self, file_path):
        if not os.path.exists(file_path):
            print(f"\nError: File '{file_path}' not found. Please check the name and try again.")
            return False
        
        try:
            self.df = pd.read_csv(file_path)
            if self.df.isnull().values.any():
                self.df.dropna(inplace=True)

            self.df['Date'] = pd.to_datetime(self.df['Date'])
            
            print(f"\nSuccess: {len(self.df)} rows loaded and validated.")
            return True
        except Exception as e:
            print(f"\nError reading CSV: {e}")
            return False

    def display_dashboard(self):
        if self.df is None:
            print("\n[!] Please load a file first (Option 1).")
            return

        print("\nOpening Dashboard... (Close the graph window to return to the menu)")
        
        plt.figure(figsize=(18, 6))
        sns.set_theme(style="whitegrid")

        plt.subplot(1, 3, 1)
        sns.barplot(data=self.df, x='Category', y='Total Sales', estimator=sum, palette='viridis', errorbar=None)
        plt.title('Total Sales by Category')
        plt.xticks(rotation=45)

        plt.subplot(1, 3, 2)
        daily_sales = self.df.groupby('Date')['Total Sales'].sum().reset_index()
        sns.lineplot(data=daily_sales, x='Date', y='Total Sales', color='red', marker='o')
        plt.title('Daily Sales Trend')
        plt.xticks(rotation=45)

        plt.subplot(1, 3, 3)
        corr = self.df[['Price', 'Quantity Sold', 'Total Sales']].corr()
        sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f")
        plt.title('Correlation: Price vs Sales')

        plt.tight_layout()
        plt.show()

    def filter_data(self):
        if self.df is None:
            print("\n[!] Please load a file first (Option 1).")
            return

        print(f"\nAvailable Categories: {list(self.df['Category'].unique())}")
        cat_input = input("Enter the Category name to filter: ").strip()

        filtered = self.df[self.df['Category'].str.lower() == cat_input.lower()]
        
        if not filtered.empty:
            print(f"\nResults for '{cat_input}' ({len(filtered)} records)")
            print(filtered.head(15))
        else:
            print(f"\nNo records found for '{cat_input}'. Check your spelling.")

if __name__ == "__main__":
    analyzer = RetailAnalyzer()
    
    while True:
        print()
        print("   RETAIL DATA ANALYZER SYSTEM")
        print()
        print("1. Load & Validate CSV File")
        print("2. View Dashboard (Bar/Line/Heatmap)")
        print("3. Filter Data by Category")
        print("4. Exit")
        
        choice = input("\nChoose an option (1-4): ")

        if choice == '1':
            filename = input("Enter file path: ")
            analyzer.load_data(filename)

        elif choice == '2':
            analyzer.display_dashboard()

        elif choice == '3':
            analyzer.filter_data()

        elif choice == '4':
            print("Have Great day. Goodbye!")
            break
        else:
            print("Invalid input. Please choose 1, 2, 3 or 4.")
```
