```python
std_rec=[]
std_UC=()
total_sub=set()
student_data={}


print("Welcome to the Student Data collector!")
while True:
    choice=input("""
Select an option:
1.  Add Student
2.  Display all Students
3.  Update Student information
4.  Delete Student
5.  Display Subjects Offered
6.  Exit

Your input: """)

    if choice=="1":
        std_id=int(input("\nEnter the student ID: "))
        if std_id in student_data:
            print("The student ID already exist")
            continue
        name=input("Enter the name of the Student: ")
        age=int(input("Enter the age of the student: "))
        grade=input("Enter the Grade of the student: ")
        sub=input("Enter the subject of the student (comma seperate the subjects): ")
        dob=input("Enter the student's D.O.B.: ")
        a=list(std_UC)
        a.append(std_id)
        a.append(dob)
        b=tuple(a)
        # print(b)
        student_data[std_id]={
            "name" : name,
            "age" : age,
            "grade" : grade,
            "sub" : sub
        }
        # print(student_data)
        print(f"\nStudent {name}'s data has successfully added.")
    
    elif choice=="2":
        for std_id in student_data:
            # print(student_data[std_id])
            print("-"*20)
            print("\nStudent ID :", std_id)
            print("Name :",student_data[std_id].get("name"))
            print("Age :",student_data[std_id].get("age"))
            print("Grade :",student_data[std_id].get("grade"))
            print("Subjects :",student_data[std_id].get("sub"))
            print("\n")
            print("-"*20)
            
    elif choice=="3":
        update=input("Enter the student ID to update the infomration of the student: ")
        print(student_data[update])
        Yes_no=input("\nIs this the student you want to update? ")
        e=Yes_no.upper()
        if e=="YES" or "Y":
            update_data=int(input("""Enter the choice of data to update
1. Age
2. Grade
3. Subjects
"""))
            if update_data==1:
                u_age=int(input("\nEnter the new age: "))
                student_data[update].update({"age" : u_age})
            if update_data==2:
                u_grade=input("\nEnter the new grade: ")
                student_data[update].update({"grade" : u_grade})
            if update_data==3:
                u_sub=input("\nEnter the new subjects: ")
                student_data[update].update({"sub" : u_sub})
        elif e=="NO" or "N":
            continue
        else:
            print("\nWrong input. type Y or N.")
            continue
        
    elif choice=="4":
        del_std=input("Enter the id of the student: ")
        print("Student name :",student_data[del_std].get("name"))
        print("Student ID:",del_std)
        print("\nDo you want to really delete the student's data? (Y/N)")
        reply=input("")
        rep=reply.upper()
        print(student_data[del_std])
        if rep=="Y":
            del student_data[del_std]
            total_sub.clear()
        elif rep=="N":
            continue
        else:
            print("Wrong input, try again!")
            continue
    
    elif choice=="5":
        for s in student_data:
            current_subject = student_data[s].get("sub")
            if current_subject:
                d = current_subject.upper()
                subject_list = [item.strip() for item in d.split(",")]
                for item in subject_list:
                    total_sub.add(item)
        print("\n" + "-"*20)
        print("TOTAL SUBJECTS OFFERED:")
        if total_sub:
            print(", ".join(total_sub))
        else:
            print("No subjects found.")
        print("-"*20)
    elif choice=="6":
        print("\nThank you for using Student DATA collector.")
        break
    else:
        print("invalid input, Try again!")
```
