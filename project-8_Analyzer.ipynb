```python
import numpy as np

class DataAnalytics:
    def __init__(self, data=None):
        self.__data = np.array(data) if data is not None else None

    def set_data(self, data):
        self.__data = np.array(data)

    def get_data(self):
        return self.__data

    def create_array(self, shape_type, dims, elements):
        arr = np.array(elements).reshape(dims)
        self.__data = arr
        return self.__data

    def index_array(self, indices):
        return self.__data[tuple(indices)]

    def slice_array(self, slices):
        return self.__data[tuple(slices)]

    def concatenate_arrays(self, other_array, axis=0):
        self.__data = np.concatenate((self.__data, other_array), axis=axis)
        return self.__data

    def split_array(self, sections, axis=0):
        return np.array_split(self.__data, sections, axis=axis)

    def element_wise_op(self, other_array, op):
        if op == 'add': return np.add(self.__data, other_array)
        if op == 'sub': return np.subtract(self.__data, other_array)
        if op == 'mul': return np.multiply(self.__data, other_array)
        if op == 'div': return np.divide(self.__data, other_array)

    def matrix_op(self, other_array, op):
        if op == 'dot': return np.dot(self.__data, other_array)
        if op == 'matmul': return np.matmul(self.__data, other_array)

    def search_value(self, value):
        return np.where(self.__data == value)

    def sort_data(self, axis=-1, descending=False):
        res = np.sort(self.__data, axis=axis)
        return res[::-1] if descending else res

    def filter_data(self, condition_str):
        return self.__data[eval(f"self._DataAnalytics__data {condition_str}")]

    def compute_aggregates(self):
        return {
            "sum": np.sum(self.__data),
            "mean": np.mean(self.__data),
            "median": np.median(self.__data),
            "std": np.std(self.__data),
            "var": np.var(self.__data)
        }

    def compute_stats(self):
        return {
            "min": np.min(self.__data),
            "max": np.max(self.__data),
            "percentiles": np.percentile(self.__data, [25, 50, 75])
        }

    @staticmethod
    def correlation(arr1, arr2):
        return np.corrcoef(arr1.flatten(), arr2.flatten())

    @classmethod
    def from_list(cls, data_list):
        return cls(np.array(data_list))

def main_menu():
    analyzer = DataAnalytics()
    while True:
        print("\nWelcome to the NumPy Analyzer!")
        print("\n1. Create a Numpy Array")
        print("2. Perform Mathematical Operations")
        print("3. Combine or Split Arrays")
        print("4. Search, Sort, or Filter Arrays")
        print("5. Compute Aggregates and Statistics")
        print("6. Exit")
        
        choice = input("Enter your choice: ")
        
        if choice == '1':
            print("""\nSelect the type of array to create:
1. 1D Array
2. 2D Array
3. 3D Array""")
            type_choice = input("Enter your choice: ")
            if type_choice == '1':
                n = int(input("Enter number of elements: "))
                els = list(map(float, input(f"Enter {n} elements: ").split()))
                analyzer.create_array("1D", (n,), els)
            elif type_choice == '2':
                r = int(input("Enter the number of rows: "))
                c = int(input("Enter the number of columns: "))
                els = list(map(float, input(f"Enter {r*c} elements: ").split()))
                analyzer.create_array("2D", (r, c), els)
            elif type_choice == '3':
                d = int(input("Enter the number of depth (layers): "))
                r = int(input("Enter the number of rows: "))
                c = int(input("Enter the number of columns: "))
                total = d * r * c
                els = list(map(float, input(f"Enter {total} elements separated by space: ").split()))
                analyzer.create_array("3D", (d, r, c), els)
            print("Array created successfully:\n", analyzer.get_data())

            print("""\nChoose an operation:
1. Indexing
2. Slicing
3. Go Back""")
            sub_choice = input("Enter your choice: ")
            if sub_choice == '2':
                r_range = input("Enter the row range (start:end): ").split(':')
                c_range = input("Enter the column range (start:end): ").split(':')
                s_obj = (slice(int(r_range[0]), int(r_range[1])), slice(int(c_range[0]), int(c_range[1])))
                print("Sliced Array:\n", analyzer.slice_array(s_obj))

        elif choice == '2':
            print("\nChoose a mathematical operation:\n1. Addition\n2. Subtraction\n3. Multiplication\n4. Division")
            op_choice = input("Enter your choice: ")
            els = list(map(float, input(f"Enter the same-size array elements: ").split()))
            other = np.array(els).reshape(analyzer.get_data().shape)
            ops = { '1': 'add', '2': 'sub', '3': 'mul', '4': 'div' }
            print("Original Array:\n", analyzer.get_data())
            print("Second Array:\n", other)
            print(f"Result:\n", analyzer.element_wise_op(other, ops[op_choice]))

        elif choice == '3':
            print("\nChoose an option:\n1. Combine Arrays\n2. Split Array")
            sub_choice = input("Enter your choice: ")
            if sub_choice == '1':
                els = list(map(float, input("Enter elements of another array: ").split()))
                other = np.array(els).reshape(analyzer.get_data().shape)
                print("Combined Array (Vertical Stack):\n", analyzer.concatenate_arrays(other, axis=0))

        elif choice == '4':
            print("\nChoose an option:\n1. Search a value\n2. Sort the array\n3. Filter values")
            sub_choice = input("Enter your choice: ")
            if sub_choice == '2':
                print("Original Array:\n", analyzer.get_data())
                print("Sorted Array:\n", analyzer.sort_data(axis=1))
                print("(Sorting applied row-wise.)")

        elif choice == '5':
            print("""\nChoose an operation:
1. Sum
2. Mean
3. Median
4. Std Dev
5. Variance""")
            op_choice = input("Enter your choice: ")
            aggs = analyzer.compute_aggregates()
            map_keys = {'1': 'sum', '2': 'mean', '3': 'median', '4': 'std', '5': 'var'}
            print(f"Original Array:\n{analyzer.get_data()}")
            print(f"{map_keys[op_choice].capitalize()} of Array: {aggs[map_keys[op_choice]]}")

        elif choice == '6':
            print("Thank you for using the NumPy Analyzer! Goodbye!")
            break

if __name__ == "__main__":
    main_menu()
```
