```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('2015.csv')

print("Missing values in dataset:")
print(df.isnull().sum())

print("\nDataset Columns:", df.columns.tolist())

top_10 = df.nlargest(10, 'Happiness Score')
print("\nTop 10 Happiest Countries in 2015:")
print(top_10[['Country', 'Happiness Score']])

region_avg = df.groupby('Region')['Happiness Score'].mean().sort_values(ascending=False).reset_index()
print("\nAverage Happiness Score by Region:")
print(region_avg)

sns.set_theme(style="whitegrid")

plt.figure(figsize=(10, 6))
sns.barplot(x='Happiness Score', y='Country', data=top_10, palette='viridis', hue='Country', legend=False)
plt.title('Top 10 Happiest Countries (2015)')
plt.xlabel('Happiness Score')
plt.ylabel('Country')
plt.tight_layout()
plt.savefig('top_10_happiness.png')

plt.figure(figsize=(10, 6))
sns.barplot(x='Happiness Score', y='Region', data=region_avg, palette='coolwarm', hue='Region', legend=False)
plt.title('Average Happiness Score by Region (2015)')
plt.xlabel('Average Happiness Score')
plt.ylabel('Region')
plt.tight_layout()
plt.savefig('happiness_by_region.png')

df_sorted = df.sort_values('Happiness Rank')

plt.figure(figsize=(10, 6))
plt.plot(df_sorted['Happiness Rank'], df_sorted['Happiness Score'], 
         color='royalblue', linewidth=2, marker='o', markersize=3, markerfacecolor='red')

plt.title('Distribution of Happiness: Score vs. Rank', fontsize=14)
plt.xlabel('Happiness Rank (1 = Happiest)', fontsize=12)
plt.ylabel('Happiness Score', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)

plt.annotate('Highest Scores', xy=(1, 7.5), xytext=(20, 7.8),
             arrowprops=dict(facecolor='black', shrink=0.05))

plt.tight_layout()
plt.savefig('score_vs_rank_line_graph.png')
print("Line graph saved as score_vs_rank_line_graph.png")

numeric_cols = df.select_dtypes(include=['float64', 'int64']).drop(columns=['Happiness Rank'])
corr_matrix = numeric_cols.corr()

plt.figure(figsize=(12, 8))
sns.heatmap(corr_matrix, annot=True, cmap='RdYlGn', fmt=".2f")
plt.title('Correlation Heatmap of Happiness Factors')
plt.tight_layout()
plt.savefig('correlation_heatmap.png')

plt.figure(figsize=(10, 6))
sns.regplot(x='Economy (GDP per Capita)', y='Happiness Score', data=df, scatter_kws={'alpha':0.5}, line_kws={"color": "red"})
plt.title('Impact of Economy (GDP per Capita) on Happiness Score')
plt.xlabel('Economy (GDP per Capita)')
plt.ylabel('Happiness Score')
plt.tight_layout()
plt.savefig('gdp_vs_happiness.png')

region_avg.to_csv('region_happiness_averages.csv', index=False)
top_10[['Country', 'Happiness Score', 'Region']].to_csv('top_10_analysis.csv', index=False)

print("\nAnalysis Complete. Plots and summary CSVs have been saved.")
```
