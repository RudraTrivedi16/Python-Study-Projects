```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

class SalesDataAnalyzer:
    def __init__(self, file_path=None):
        self.data = None
        if file_path:
            self.load_data(file_path)

    def __del__(self):
        pass

    def load_data(self, file_path):
        try:
            self.data = pd.read_csv(file_path)
            return True
        except Exception as e:
            print(f"Error loading file: {e}")
            return False

    def explore_data(self):
        if self.data is None: return
        print("\n--- Basic Info ---")
        self.data.info()
        print("\n--- First 5 Rows ---")
        print(self.data.head())

    def clean_data(self):
        if self.data is None: return
        self.data.drop_duplicates(inplace=True)
        numeric_cols = self.data.select_dtypes(include=[np.number]).columns
        self.data[numeric_cols] = self.data[numeric_cols].fillna(self.data[numeric_cols].mean())
        self.data.fillna("Unknown", inplace=True)

    def mathematical_operations(self):
        if self.data is None: return
        sales_array = self.data['Sales'].to_numpy()
        taxed_sales = sales_array * 1.15
        print(f"Original first 5: {sales_array[:5]}")
        print(f"Taxed first 5: {taxed_sales[:5]}")

    def combine_data(self, other_df):
        if self.data is not None:
            self.data = pd.concat([self.data, other_df], ignore_index=True)

    def split_data(self, column):
        if self.data is None: return {}
        return {val: df for val, df in self.data.groupby(column)}

    def search_sort_filter(self):
        if self.data is None: return
        filtered = self.data[self.data['Sales'] > 500]
        sorted_df = filtered.sort_values(by='Sales', ascending=False)
        return sorted_df

    def aggregate_functions(self):
        if self.data is None: return
        return self.data.aggregate({'Sales': ['sum', 'mean', 'count'], 'Profit': ['sum', 'mean']})

    def statistical_analysis(self):
        if self.data is None: return
        return self.data.describe()

    def create_pivot_table(self):
        if self.data is None: return
        return pd.pivot_table(self.data, values='Sales', index='Region', columns='Year', aggfunc='sum')

    def visualize_data(self, chart_type, x=None, y=None, save_path=None):
        if self.data is None: return
        plt.figure(figsize=(10, 6))

        if chart_type == 1:
            sns.barplot(data=self.data, x=x, y=y)
        elif chart_type == 2:
            sns.lineplot(data=self.data, x=x, y=y)
        elif chart_type == 3:
            sns.scatterplot(data=self.data, x=x, y=y)
        elif chart_type == 4:
            self.data[x].value_counts().plot.pie(autopct='%1.1f%%')
        elif chart_type == 5:
            sns.histplot(self.data[x], kde=True)

        plt.title(f"{x} vs {y}" if y else f"Distribution of {x}")

        if save_path:
            plt.savefig(save_path)
            print(f"Saved to {save_path}")
        plt.show()

def main_menu():
    analyzer = SalesDataAnalyzer()

    while True:
        print("\n========== Data Analysis & Visualization Program ==========")
        print("1. Load Dataset")
        print("2. Explore Data")
        print("3. Perform DataFrame Operations")
        print("4. Handle Missing Data")
        print("5. Generate Descriptive Statistics")
        print("6. Data Visualization")
        print("7. Save Visualization")
        print("8. Exit")

        choice = input("\nEnter your choice: ")

        if choice == '1':
            path = input("Enter path: ")
            if analyzer.load_data(path):
                print("Dataset loaded successfully!")

        elif choice == '2':
            print("\n1. Display first 5 rows\n2. Display basic info")
            sub = input("Choice: ")
            if sub == '1': print(analyzer.data.head())
            else: analyzer.explore_data()

        elif choice == '4':
            analyzer.clean_data()
            print("Data cleaned (Duplicates dropped, missing values filled).")

        elif choice == '5':
            print(analyzer.statistical_analysis())

        elif choice == '6':
            print("\n1. Bar\n2. Line\n3. Scatter\n4. Pie\n5. Histogram")
            c = int(input("Chart choice: "))
            x_col = input("X axis column: ")
            y_col = input("Y axis column (if applicable): ")
            analyzer.visualize_data(c, x_col, y_col)

        elif choice == '8':
            print("Exiting the program. Goodbye!")
            break

if __name__ == "__main__":
    main_menu()
```
