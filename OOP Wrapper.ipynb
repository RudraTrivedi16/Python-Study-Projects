```python
class Employee:
    def __init__(self, employee_id=None, name=None, age=None, salary=None):
        self.__employee_id = employee_id
        self.name = name
        self.age = age
        self.__salary = salary

    def get_employee_id(self):
        return self.__employee_id
    
    def set_employee_id(self, employee_id):
        self.__employee_id = employee_id

    def get_salary(self):
        return self.__salary
    
    def set_salary(self, salary):
        self.__salary = salary

    def display(self):
        print(f"ID: {self.__employee_id}, Name: {self.name}, Age: {self.age}, Salary: {self.__salary}")

    def __del__(self):
        print(f"Employee {self.name} with ID {self.__employee_id} is deleted.")

class Manager(Employee):
    def __init__(self, employee_id, name, age, salary, department):
        super().__init__(employee_id, name, age, salary)
        self.department = department

    def display(self):
        super().display()
        print(f"Department: {self.department}")

class Developer(Employee):
    def __init__(self, employee_id, name, age, salary, programming_language):
        super().__init__(employee_id, name, age, salary)
        self.programming_language = programming_language

    def display(self):
        super().display()
        print(f"Programming Language: {self.programming_language}")

def main():
    employees = {}

    while True:
        print("\nWelcome to Employee Management System. Choose from following options.")
        print("1. Add Employee")
        print("2. Update Employee Salary")
        print("3. Remove Employee")
        print("4. Display Employee")
        print("5. Exit")

        choice = input("\nEnter your choice: ")

        if choice == "1":
            print("Choose type: 1. Manager  2. Developer  3. Employee")
            emp_type = input("Enter type: ")

            emp_id = input("Enter Employee ID: ")
            name = input("Enter Name: ")
            age = int(input("Enter Age: "))
            salary = float(input("Enter Salary: "))

            if emp_type == "1":
                dept = input("Enter Department: ")
                employees[emp_id] = Manager(emp_id, name, age, salary, dept)
            elif emp_type == "2":
                lang = input("Enter Programming Language: ")
                employees[emp_id] = Developer(emp_id, name, age, salary, lang)
            else:
                employees[emp_id] = Employee(emp_id, name, age, salary)

            print("\nEmployee added successfully.")

        elif choice == "2":
            emp_id = input("\nEnter Employee ID to update salary: ")
            if emp_id in employees:
                new_salary = float(input("Enter new salary: "))
                employees[emp_id].set_salary(new_salary)
                print("\nSalary updated successfully.")
            else:
                print("\nEmployee not found.")

        elif choice == "3":
            emp_id = input("\nEnter Employee ID to remove: ")
            if emp_id in employees:
                del employees[emp_id]
                print("\nEmployee removed successfully.")
            else:
                print("\nEmployee not found.")

        elif choice == "4":
            emp_id = input("\nEnter Employee ID to display: ")
            if emp_id in employees:
                print("")
                employees[emp_id].display()
            else:
                print("\nEmployee not found.")

        elif choice == "5":
            print("\nThank you for using the Employee Management System.")
            break

        else:
            print("\nInvalid input. Try again.")


if __name__ == "__main__":
    main()
```
