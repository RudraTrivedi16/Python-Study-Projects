```python
import os
from datetime import datetime

class JournalManager:
    def __init__(self, filename="journal.txt"):
        self.filename = filename

    def add_entry(self, text):
        try:
            with open(self.filename, "a") as file:
                timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
                file.write(f"{timestamp} - {text}\n")
            print("\nEntry added successfully!")
        except PermissionError:
            print("\nError: Permission not granted")
 
    def view_entries(self):
        try:
            with open(self.filename, "r") as file:
                content = file.read()
                if content.strip() == "":
                    print("\nJournal is empty.")
                else:
                    print("\n* Journal Entries --")
                    print(content)
        except FileNotFoundError:
            print("\nError: Journal file does not exist.")

    def search_entries(self, keyword):
        try:
            with open(self.filename, "r") as file:
                lines = file.readlines()
                matches = [line for line in lines if keyword.lower() in line.lower()]
                if matches:
                    print("\n* Search Results")
                    for match in matches:
                        print(match.strip())
                else:
                    print("\nNo entries found with that keyword.")
        except FileNotFoundError:
            print("\nError: Journal file does not exist yet.")

    def delete_entries(self):
        if os.path.exists(self.filename):
            confirm = input("\nAre you sure you want to delete all entries? (y/n): ")
            if confirm.lower() == "y":
                os.remove(self.filename)
                print("\nAll entries deleted.")
            else:
                print("\nDelete cancelled.")
        else:
            print("\nJournal file does not exist.")


def main():
    jm = JournalManager()

    while True:
        print("\nPersonal Journal :")
        print("\n1. Add a New Entry")
        print("2. View All Entries")
        print("3. Search for an Entry")
        print("4. Delete All Entries")
        print("5. Exit")

        choice = input("\nEnter your choice: ")

        if choice == "1":
            text = input("\nWrite your entry: ")
            jm.add_entry(text)
        elif choice == "2":
            jm.view_entries()
        elif choice == "3":
            keyword = input("\nEnter keyword to search: ")
            jm.search_entries(keyword)
        elif choice == "4":
            jm.delete_entries()
        elif choice == "5":
            print("\nGoodbye!")
            break
        else:
            print("\nInvalid choice. Try again.")


if __name__ == "__main__":
    main()
```
